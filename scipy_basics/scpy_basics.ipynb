## optimizers = use to find root for non linear,linear equation and helpful for minimze the data

```python
## optimizers
from scipy.optimize import root
from math import cos
def eqn(x):
    return x + cos(x)
my_root = root(eqn,0)
print(my_root)
print()
print(my_root.x)        ## .x because we want only result not the various calculation inside the solution
```

```python
## minimize
from scipy.optimize import minimize
def eqn(x):
    return x**2 + x + 2
my_min = minimize(eqn,0,method = 'BFGS')    ## single quotes always
print(my_min)
print()
print(my_min.x)
```

## sparse data:

## dense array: opposite of sparse data

```python
from scipy.sparse import csr_matrix
import numpy as np
arr = np.array([0,0,0,0,0,2,0,0,1,3,0,4])
print(csr_matrix(arr))
```

```python
print(csr_matrix(arr).data)
print(csr_matrix(arr).count_nonzero())
```

```python
arr = np.array([0,0,0,0,0,2,0,0,1,3,0,4])
mat = csr_matrix(arr)
mat.eliminate_zeros()
print(mat)
```

```python
arr = np.array([0,0,0,0,0,2,0,0,1,3,0,4])
mat = csr_matrix(arr)
mat.sum_duplicates()
print(mat)
```

## graphs to store data
1 connected components
2 djikstra
3 floyd worshall
4 bellman ford
5 depth first order
6 breadth first order

```python
from scipy.sparse.csgraph import connected_components
from scipy.sparse import csr_matrix
arr = np.array([[1,0,2],[1,0,0],[0,0,2]])
newarr = csr_matrix(arr)
print(connected_components(newarr))
```

```python
from scipy.sparse.csgraph import dijkstra
from scipy.sparse import csr_matrix
arr = np.array([[1,0,2],[1,0,0],[0,0,2]])
newarr = csr_matrix(arr)
print(dijkstra(newarr,return_predecessors = True,indices = 0))
```

```python
from scipy.sparse.csgraph import floyd_warshall
from scipy.sparse import csr_matrix
arr = np.array([[1,0,2],[1,0,0],[0,0,2]])
newarr = csr_matrix(arr)
print(floyd_warshall(newarr,return_predecessors = True))
```

```python
from scipy.sparse.csgraph import bellman_ford
from scipy.sparse import csr_matrix
arr = np.array([[1,0,2],[1,0,0],[0,0,2]])
newarr = csr_matrix(arr)
print(bellman_ford(newarr,return_predecessors = True,indices = 0))
```

```python
from scipy.sparse.csgraph import depth_first_order
from scipy.sparse import csr_matrix
arr = np.array([[1,0,2],[1,0,0],[0,0,2]])
newarr = csr_matrix(arr)
print(depth_first_order(newarr,1))
```

```python
from scipy.sparse.csgraph import breadth_first_order
from scipy.sparse import csr_matrix
arr = np.array([[1,0,2],[1,0,0],[0,0,2]])
newarr = csr_matrix(arr)
print(breadth_first_order,(newarr,1))
```

## distance 
1 euclidean
2 cityblock
3 cosine
4 hamming


```python
from scipy.spatial.distance import euclidean
p1 =(2,6)
p2 = (5,2)
res = euclidean(p1,p2)
print(res)
```

```python
from scipy.spatial.distance import cityblock
p1 =(2,6)
p2 = (5,2)
res = cityblock(p1,p2)
print(res)
```

```python
from scipy.spatial.distance import cosine
p1 =(2,6)
p2 = (5,2)
res = cosine(p1,p2)
print(res)
```

```python
from scipy.spatial.distance import hamming
p1 =(True,False,True,True)
p2 = (False,True,True,False)
res = hamming(p1,p2)
print(res)
```

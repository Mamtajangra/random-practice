```python
import re
def remove_html_tags(text):
    pattern = re.compile(r"<.*?>")
    return pattern.sub("", text) 
```

# remove url

```python
def remove_url(text):
    pattern = re.compile(r'https?://\S+|www\.\S+')
    return pattern.sub(r"",text)
```

```python
text1 = "check out my notebook https://hina.com"
text2 = "check out my notebook https://dummy-domain.net"
text3 = "check out my notebook https://testsite.org"
text4 = "check out my notebook https://example.com"
```

```python
remove_url(text1)
```

```python
remove_url(text4)
```

# remove punctuation
(because during tokenization these marks converted into numbers and resulted increase the size of data so that require to remove)

```python
import string
string.punctuation
```

```python
exclude = string.punctuation
```

```python
def remove_punct(text):
    return text.translate(str.maketrans("","",exclude))
```

```python
text = "Wait, what?! exclaimed Dr. Smith-Jones—the lead NLP scientist at AI_Labs (a subsidiary of TechCorp)." 
```

```python
remove_punct(text)
```

```python
df["review"].apply(remove_punct)
```

```python
def chat_conversion(text):
    new_text = []
    for w in text.split():
        if w.lower() in chat_words:
            new_text.append(chat_words[w.lower()])
        else:
            new_text.append(w)
    return ' '.join(new_text)            
```

# removal stopwards
(and,are,or....etc)

```python
from nltk.corpus import stopwords
```

```python
stopwords.words("english")
```

```python
def remove_stopwords(text):
    new_text = []
    for word in text.split():
        if word in stopwords.words("english"):
            new_text.append("")
        else:
            new_text.append(word)
    x = new_text[:]
    new_text.clear()
    return" ".join(x)            
```

```python
text = "sita and geeta are two sisters they love each other but one s tall and another is short so jealousy occur between them"
remove_stopwords(text)
```

```python
df["review"].apply(remove_stopwords)
```
